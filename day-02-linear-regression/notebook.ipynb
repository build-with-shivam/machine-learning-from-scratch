```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
from sklearn.datasets import fetch_california_housing
from matplotlib import pyplot as plt

print("all modules imported successfully")
```

```python
# dataset
housing = fetch_california_housing()

# dataframe
x = pd.DataFrame(housing.data, columns=housing.feature_names)
y = pd.Series(housing.target, name='target')
print("Dataframe created successfully")
print(f"Shape of dataframe: {x.shape}")
print(f"Shape of target: {y.shape}")

# explore dataset
print("Exploring dataset...")
print(x.head())
print(x.describe())
print(x.info())
print(y.head())
print(y.describe())
```

```python
# test train split

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
print("\nDataset Split:")
print(f"Training samples: {len(X_train)}")
print(f"Testing samples : {len(X_test)}")
```

```python
# create and train model

print("\nCreating and training Linear Regression model...")
model = LinearRegression()
model.fit(X_train, y_train)
print("Model Trained Successfully")
```

```python
# make predictions

print("\nMaking predictions on test set...")
y_pred = model.predict(X_test)
print("Predictions made successfully")
```

```python
# Evaluate model

mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = mse ** 0.5

r2 = r2_score(y_test, y_pred)


print("\nModel Performance")
print("-" * 30)
print(f"MAE  : {mae:.4f}")
print(f"MSE  : {mse:.4f}")
print(f"RMSE : {rmse:.4f}")
print(f"R²   : {r2:.4f}")
```

```python
# display sample predictions

results = pd.DataFrame(
    {
        "Actual": y_test.values,
        "Predicted": y_pred,
    }
)

print("\nSample Predictions:")
print(results.head(10))
```

```python
# visualize results

plt.figure(figsize=(8, 6))

plt.scatter(
    y_test,
    y_pred,
    alpha=0.5,
)

plt.xlabel("Actual House Price")
plt.ylabel("Predicted House Price")
plt.title("Actual vs Predicted House Prices")

# Perfect prediction reference line
min_value = min(y_test.min(), y_pred.min())
max_value = max(y_test.max(), y_pred.max())

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    linestyle="--",
)

plt.tight_layout()

plot_file = OUTPUT_DIR / "actual_vs_predicted.png"

plt.savefig(plot_file)

plt.show()

print(f"Plot saved to: {plot_file}")


print("\nHouse Price Prediction completed successfully! 🚀")
```
